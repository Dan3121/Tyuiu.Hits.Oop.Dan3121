**Проект для управления растениями**

**Спринт ревью #0**

```
using System;

public class Plant
{
    public string Name { get; set; }
    public string Type { get; set; }
    public float Height { get; set; } 

    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual void Bloom()
    {
        Console.WriteLine($"{Name} сейчас цветет ");
    }

    public void Grow(float growth)
    {
        Height += growth;
        Console.WriteLine($"{Name} выросло на {growth} см. текущая высота: {Height} см");
    }
}

public class Flower : Plant
{

    public string Color { get; set; }
    public int Lepestki_Count { get; set; }

    public Flower(string name, float height, string color, int lepestki_Count) 
    : base(name, "Цветок", height)
    {
        Color = color;
        Lepestki_Count = lepestki_Count;
    }

    public override void Bloom()
    {
        Console.WriteLine($"цветок {Name} распускается и радует глаз");
    }

    public void Wither()
    {
        Console.WriteLine($"цветок {Name} начинает увядать");
    }
}

public class Tree : Plant
{
    public int Age { get; set; }
    public string Leaf_Type { get; set; }

    public Tree(string name, float height, int age, string leaf_Type) 
        : base(name, "дерево", height)
    {
        Age = age;
        Leaf_Type = leaf_Type;
    }
    public override void Bloom()
    {
        Console.WriteLine($"дерево {Name} покрывается новыми листьями");
    }

    public void GrowAge(int years)
    {
        Age += years;
        Console.WriteLine($"дерево {Name} подросло на {years} год(а). текущий возраст: {Age} лет");
    }
}

Flower rose = new Flower("Роза", 25.0f, "Красный", 15);
Tree oak = new Tree("Дуб", 300.0f, 10, "Лиственные");

Console.WriteLine("полиморфизм (Цветение)");
rose.Bloom();
oak.Bloom();

Console.WriteLine("(Рост)");
rose.Grow(5.5f);
oak.Grow(20.0f);

Console.WriteLine("(Разное что-то)");
rose.Wither();
oak.GrowAge(2);

Console.WriteLine("полиморфизм(Plant)");
Plant[] garden = new Plant[] { rose, oak };
foreach (Plant plant in garden)
{
    plant.Bloom();
}
```
